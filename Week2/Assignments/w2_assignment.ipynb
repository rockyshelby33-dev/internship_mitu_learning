##  Question 1: analyze_sentence()

```python
def analyze_sentence(sentence):
    words = sentence.split()
    word_count = len(words)
    char_count = len(sentence.replace(" ", ""))
    unique_words = set(word.lower() for word in words)
 
    # Count frequency of each word (case-insensitive)
    freq = {}
    for word in words:
        key = word.lower()
        freq[key] = freq.get(key, 0) + 1
 
    most_frequent_word = max(freq, key=freq.get)
 
    return {
        "word_count": word_count,
        "char_count": char_count,
        "unique_words": unique_words,
        "most_frequent_word": most_frequent_word
    }
 
result = analyze_sentence("The quick brown fox jumps over the lazy dog the fox")
for key, value in result.items():
    print(f"  {key}: {value}")
 
```

## Question 2: String Operations on text

```python
text = "A man a plan a canal Panama"
# a) 
reversed_text = ""
for char in text:
    reversed_text = char + reversed_text
print(f"\na) Reversed           : {reversed_text}")
 
# b) 
cleaned = text.replace(" ", "").lower()
is_palindrome = (cleaned == cleaned[::-1])
print(f"b) Is Palindrome?     : {is_palindrome}")
 
# c) 
vowels = "aeiouAEIOU"
replaced = ""
for char in text:
    replaced += "*" if char in vowels else char
print(f"c) Vowels replaced    : {replaced}")
 
# d) 
vowel_count = {}
for char in text.lower():
    if char in "aeiou":
        vowel_count[char] = vowel_count.get(char, 0) + 1
print(f"d) Vowel counts       : {vowel_count}")
 
```

##  Question 4: list_ops()

```python
def list_ops(nums):
    # a) Remove duplicates, preserve order
    seen = []
    for n in nums:
        if n not in seen:
            seen.append(n)
    no_duplicates = seen
 
    # b) Elements at even indices, squared
    even_index_squared = [nums[i] ** 2 for i in range(0, len(nums), 2)]
 
    return no_duplicates, even_index_squared
 
nums   = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]
nested = [[1, 2, 3], [4, 5], [6, 7, 8, 9]]
 
no_dups, even_sq = list_ops(nums)
print(f"\nOriginal              : {nums}")
print(f"a) No duplicates      : {no_dups}")
print(f"b) Even-index squared : {even_sq}")
 
# c) Flatten nested list using list comprehension
flattened = [item for sublist in nested for item in sublist]
print(f"c) Flattened nested   : {flattened}")
 
```

##  Question 7: inventory_manager()

```python
def inventory_manager(records):
    inventory = {}
    for product, category, quantity, price in records:
        if category not in inventory:
            inventory[category] = {
                "total_items" : 0,
                "total_value" : 0,
                "products"    : []
            }
        inventory[category]["total_items"] += quantity
        inventory[category]["total_value"] += quantity * price
        inventory[category]["products"].append(product)
    return inventory
 
records = [
    ("Laptop", "Electronics", 10, 55000),
    ("Mouse",  "Electronics", 50, 499),
    ("Desk",   "Furniture",    8, 12000),
    ("Chair",  "Furniture",   15, 4500),
    ("Pen",    "Stationery", 200, 20),
]
 
inv = inventory_manager(records)
print()
for category, details in inv.items():
    print(f"Category     : {category}")
    print(f"  Total Items: {details['total_items']}")
    print(f"  Total Value: ₹{details['total_value']:,}")
    print(f"  Products   : {details['products']}")
    print()
 
```

## Question 8: *args, **kwargs, lambda

```python
# a) power_all using *args
def power_all(*args, exp=2):
    return [n ** exp for n in args]
 
print(f"\na) power_all(2,3,4)        : {power_all(2, 3, 4)}")
print(f"   power_all(2,3,4,exp=3)  : {power_all(2, 3, 4, exp=3)}")
 
# b) profile using **kwargs
def profile(**kwargs):
    print()
    for key, value in kwargs.items():
        print(f"   {key} : {value}")
 
print("\nb) profile() output:")
profile(name="Shreyas", role="Data Science Intern", city="Mumbai")
 
# c) Lambda inside sorted() — sort by cgpa descending
students = [
    {"name": "Arjun", "cgpa": 8.4},
    {"name": "Sanya", "cgpa": 9.1},
    {"name": "Dev",   "cgpa": 7.8},
]
sorted_students = sorted(students, key=lambda s: s["cgpa"], reverse=True)
print("\nc) Sorted by CGPA (descending):")
for s in sorted_students:
    print(f"   {s['name']:<10} → CGPA: {s['cgpa']}")
 
 
```

##  Question 9: stats_utils module + main

```python
# Run THIS cell first ↓
def mean(data):
    return sum(data) / len(data)

def variance(data):
    mu = mean(data)
    return sum((x - mu) ** 2 for x in data) / len(data)

def normalize(data):
    min_val = min(data)
    max_val = max(data)
    return [(x - min_val) / (max_val - min_val) for x in data]
```

```python
import math
dataset = [12, 45, 7, 23, 56, 89, 34, 67, 11, 90]
mu         = mean(dataset)
var        = variance(dataset)
std_dev    = math.sqrt(var)        # std dev using math module
normalized = normalize(dataset)

print(f"Dataset    : {dataset}")
print(f"Mean       : {mu:.2f}")
print(f"Variance   : {var:.2f}")
print(f"Std Dev    : {std_dev:.2f}  (via math.sqrt)")
print(f"Normalized : {[round(v, 3) for v in normalized]}")
 
```

## Question 10: Closures & Decorators

```python
# a) Closure — make_multiplier(n)
def make_multiplier(n):
    def multiplier(x):
        return x * n
    return multiplier
 
triple    = make_multiplier(3)
quintuple = make_multiplier(5)
 
print(f"\na) triple(7)    = {triple(7)}")
print(f"   quintuple(7) = {quintuple(7)}")
 
# b) @timer decorator
def timer(func):
    def wrapper(*args, **kwargs):
        start  = time.time()
        result = func(*args, **kwargs)
        end    = time.time()
        print(f"   ⏱ {func.__name__}() took {end - start:.6f} seconds")
        return result
    return wrapper
 
@timer
def slow_sum(n):
    return sum(i ** 2 for i in range(1, n + 1))
 
print(f"\nb) slow_sum(10000) = {slow_sum(10000)}")
 
# c) @validate_positive decorator
def validate_positive(func):
    def wrapper(*args, **kwargs):
        for arg in args:
            if arg <= 0:
                raise ValueError(
                    f"Invalid input: {arg}. All arguments must be positive."
                )
        return func(*args, **kwargs)
    return wrapper
 
@validate_positive
def safe_sqrt(x):
    return math.sqrt(x)
 
print(f"\nc) safe_sqrt(25) = {safe_sqrt(25)}")
print(f"   safe_sqrt(2)  = {safe_sqrt(2):.6f}")
 
# Demonstrate the error
try:
    safe_sqrt(-9)
except ValueError as e:
    print(f"   safe_sqrt(-9) → ValueError: {e}")
 
```
