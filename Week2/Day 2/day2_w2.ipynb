```python
l = [34, 56, 12, 48, 19, 34, 10]
```

```python
for index in range(len(l)):
    value = l[index] 
    print(f"{index} => {value}")
```

```python
for i in l:
    print(f"{l.index(i)} => {i}")
    ##using index
```

```python
## using Enumerate
```

```python
list(enumerate(l))
```

```python
for ind, val in enumerate(l):
    print(f"{ind} => {val}")
```

```python
list(enumerate(l, start=101))
```
