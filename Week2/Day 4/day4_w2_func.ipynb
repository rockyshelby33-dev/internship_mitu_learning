```python
# function to find the factorial of a num
```

```python
def fact(): #ex
    num = int(input("Enter a Num: "))
    
    # Starting tracker at 1 (since multiplying by 0 results in 0)
    c = 1 
    
    # range(1, num + 1) generates numbers from 1 right up to your target number
    for i in range(1, num + 1):
        c = c * i  # Multiply the current number into our running total
        
    print(f"The factorial of {num} is {c}")
```

```python
fact()
```

```python
fact()
```

```python
def find_total_sum(*args: int):
    """
    Accepts variable-length integer arguments 
    and returns their accumulated sum.
    """
    total_sum = 0
    
    # Python lets us step directly through the args tuple
    for num in args:
        total_sum += num  # Accumulating the values
        
    return total_sum

```

```python
print(find_total_sum(10, 20, 30))
print(find_total_sum(5, 1, 4, 8, 2))    
```
