```python
user_string = input("Enter the main string: ")
char1 = input("Enter Char1 to search: ")
char2 = input("Enter Char2 to search: ")

# Put both characters into a list so we can check them in a loop
targets = [char1, char2]

# Outer loop: Look at Char1 first, then look at Char2
for target in targets:
    
    found = False  # Reset our tracker flag for each character
    
    # Inner loop: Check every single letter in the main string
    for letter in user_string:
        if letter == target:
            found = True
            break  # Match found! Stop scanning the rest of the letters
            
    # If/Else block to handle the final output for the current target
    if found == True:
        print(f"-> '{target}' is PRESENT")
    else:
        print(f"-> '{target}' is NOT PRESENT")
```
