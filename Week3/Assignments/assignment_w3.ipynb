```python
# Q1. BankAccount

class BankAccount:
    def __init__(self, account_holder, account_number, balance):
        self.account_holder = account_holder
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance!")
            return
        self.balance -= amount

    def get_balance(self):
        return self.balance

    def __str__(self):
        return f"Holder: {self.account_holder}, Acc No: {self.account_number}, Balance: {self.balance}"


acc1 = BankAccount("Shreyas", 1001, 5000)
acc2 = BankAccount("Johan", 1002, 3000)

acc1.deposit(1000)
acc1.withdraw(2000)

acc2.deposit(500)
acc2.withdraw(4000)

print(acc1)
print(acc2)
```

```python
# Q2. Student

class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def calculate_percentage(self):
        return sum(self.marks) / len(self.marks)

    def get_grade(self):
        per = self.calculate_percentage()

        if per >= 80:
            return "A"
        elif per >= 60:
            return "B"
        elif per >= 40:
            return "C"
        else:
            return "Fail"

    def display(self):
        print(f"Name: {self.name}")
        print(f"Roll No: {self.roll_no}")
        print(f"Marks: {self.marks}")
        print(f"Percentage: {self.calculate_percentage():.2f}")
        print(f"Grade: {self.get_grade()}")
        print("-" * 30)


students = [
    Student("A", 1, [80, 85, 90, 75, 88]),
    Student("B", 2, [60, 65, 70, 58, 62]),
    Student("C", 3, [35, 40, 38, 42, 30])
]

for s in students:
    s.display()
```

```python
# Q3. Employee (Encapsulation)

class Employee:
    def __init__(self, name, emp_id, salary):
        self.__name = name
        self.__emp_id = emp_id
        self.__salary = salary

    def get_name(self):
        return self.__name

    def set_name(self, name):
        self.__name = name

    def get_emp_id(self):
        return self.__emp_id

    def set_emp_id(self, emp_id):
        self.__emp_id = emp_id

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary < 10000:
            raise ValueError("Salary must be at least 10000")
        self.__salary = salary

    def give_raise(self, percent):
        self.__salary += self.__salary * percent / 100


emp = Employee("Rahul", 101, 20000)
emp.give_raise(10)

print(emp.get_name())
print(emp.get_salary())
```

```python
# Q4. SecureVault

class SecureVault:
    def __init__(self, pin):
        self.__pin = pin
        self.__data = {}

    def add_entry(self, pin, key, value):
        if pin == self.__pin:
            self.__data[key] = value
        else:
            print("Access Denied")

    def get_entry(self, pin, key):
        if pin == self.__pin:
            return self.__data.get(key)
        else:
            print("Access Denied")


vault = SecureVault("1234")
vault.add_entry("1234", "email", "abc@gmail.com")
print(vault.get_entry("1234", "email"))
vault.get_entry("9999", "email")
```

```python
import math
# Q5. Shape Hierarchy

class Shape:
    def area(self):
        return 0

    def describe(self):
        print("I am a shape")


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def describe(self):
        print("I am a circle")


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def describe(self):
        print("I am a rectangle")


class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height

    def describe(self):
        print("I am a triangle")


c = Circle(5)
r = Rectangle(4, 6)
t = Triangle(10, 8)

c.describe()
print(c.area())

r.describe()
print(r.area())

t.describe()
print(t.area())
```

```python
# Q6. Vehicle

class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def start_engine(self):
        print("Engine started")


class Car(Vehicle):
    def __init__(self, make, model, year, num_doors):
        super().__init__(make, model, year)
        self.num_doors = num_doors

    def start_engine(self):
        print(f"{self.make} {self.model} engine started")


class ElectricCar(Vehicle):
    def __init__(self, make, model, year, battery_capacity):
        super().__init__(make, model, year)
        self.battery_capacity = battery_capacity

    def start_engine(self):
        print(f"Silent start — battery at {self.battery_capacity} kWh")


car = Car("Toyota", "Camry", 2023, 4)
ev = ElectricCar("Tesla", "Model 3", 2024, 75)

car.start_engine()
ev.start_engine()
```

```python
# Q7. Polymorphism


class Animal:
    def sound(self):
        print("Some sound")


class Dog(Animal):
    def sound(self):
        print("Bark")


class Cat(Animal):
    def sound(self):
        print("Meow")


class Cow(Animal):
    def sound(self):
        print("Moo")


def make_all_sounds(animals):
    for animal in animals:
        animal.sound()


animals = [Dog(), Cat(), Cow()]
make_all_sounds(animals)

```

```python

# Q8. Calculator & Vector

class Calculator:
    def add(self, a, b, c=0):
        return a + b + c

    def multiply(self, a, b, c=1):
        return a * b * c


calc = Calculator()

print(calc.add(2, 3))
print(calc.add(2, 3, 4))

print(calc.multiply(2, 3))
print(calc.multiply(2, 3, 4))


class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"


v1 = Vector(2, 3)
v2 = Vector(4, 5)

print(v1 + v2)

```

```python
# Q10. Student Record Manager


class StudentManager:
    def __init__(self):
        self.students = []

    def add_student(self):
        name = input("Name: ")
        roll = input("Roll No: ")

        marks = []
        for i in range(3):
            marks.append(float(input(f"Subject {i+1}: ")))

        self.students.append({
            "name": name,
            "roll": roll,
            "marks": marks
        })

    def view_all(self):
        for s in self.students:
            avg = sum(s["marks"]) / 3
            print(s["name"], s["roll"], avg)

    def search_student(self):
        roll = input("Enter Roll No: ")

        for s in self.students:
            if s["roll"] == roll:
                print(s)

    def delete_student(self):
        roll = input("Enter Roll No: ")

        self.students = [
            s for s in self.students
            if s["roll"] != roll
        ]

    def save_to_file(self):
        with open("students_data.txt", "w") as file:
            for s in self.students:
                file.write(f"{s['name']},{s['roll']},{s['marks']}\n")

    def load_from_file(self):
        try:
            with open("students_data.txt", "r") as file:
                print("Data Loaded")
        except FileNotFoundError:
            pass


manager = StudentManager()
manager.load_from_file()

while True:
    print("\n1.Add")
    print("2.View")
    print("3.Search")
    print("4.Delete")
    print("5.Save")
    print("6.Load")
    print("7.Exit")

    choice = input("Enter choice: ")

    if choice == "1":
        manager.add_student()

    elif choice == "2":
        manager.view_all()

    elif choice == "3":
        manager.search_student()

    elif choice == "4":
        manager.delete_student()

    elif choice == "5":
        manager.save_to_file()

    elif choice == "6":
        manager.load_from_file()

    elif choice == "7":
        break

    else:
        print("Invalid choice")




```
