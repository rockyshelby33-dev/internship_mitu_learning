```python
class Employee:
    company = "TechCorp"

    def __init__(self, name, monthly_salary, job_designation):
        self.name = name
        self.job_designation = job_designation
        self.__monthly_salary = monthly_salary
    def calculate_annual_salary(self):
        return self.__monthly_salary * 12
    def show_data(self):
        print(f"Company: {Employee.company}")
        print(f"Name: {self.name}")
        print(f"Designation: {self.job_designation}")
        print(f"Monthly Salary: ${self.__monthly_salary}")
        print(f"Annual Salary: ${self.calculate_annual_salary()}")
        print("-" * 30)

e1 = Employee("Alice Smith", 5000, "Software Engineer")
e2 = Employee("Bob Jones", 620, "Data Scientist")
e3 = Employee("Charlie Brown", 4500, "QA Analyst")

e1.show_data()
e2.show_data()
e3.show_data()
```

```python
# Manager subclass that inherits from Employee
class Manager(Employee):
    """
    Manager inherits from Employee.
    Managers have all Employee attributes + team management features.
    """
    
    def __init__(self, name, monthly_salary, job_designation, team_size):
        # Call parent class constructor to initialize employee attributes
        super().__init__(name, monthly_salary, job_designation)
        self.team_size = team_size
    
    def calculate_bonus(self, bonus_percentage=15):
        """
        Managers get a higher default bonus (15%) plus a team bonus.
        Why override? Managers deserve higher bonuses based on team performance.
        """
        base_bonus = super().calculate_bonus(bonus_percentage)
        team_bonus = self.team_size * 500  # $500 per team member
        return base_bonus + team_bonus
    
    def show_data(self):
        """Override to show additional manager info"""
        super().show_data()  # Call parent's show_data first
        print(f"Team Size: {self.team_size} employees")
        print(f"Total Bonus (with team bonus): ${self.calculate_bonus():.2f}")
        print("-" * 30)

# Create Manager instances
m1 = Manager("Diana Prince", 7000, "Engineering Manager", team_size=5)
m2 = Manager("Edward Norton", 6500, "Product Manager", team_size=3)

print("=== MANAGERS ===")
m1.show_data()
m2.show_data()

print("\n=== COMPARISON ===")
print(f"Regular Employee (Alice) bonus: ${e1.calculate_bonus():.2f}")
print(f"Manager (Diana) bonus: ${m1.calculate_bonus():.2f}")
print(f"\nWhy the difference? Diana manages 5 people + gets higher base bonus!")
```

## Inheritance: Manager Subclass

**Why inheritance?**
- A Manager IS-AN Employee, so they share the same attributes and methods
- But Managers have extra responsibilities: managing a team
- **DRY Principle**: Don't Repeat code — reuse Employee's logic
- The `super()` keyword lets us call the parent class's `__init__`

```python
# Enhanced Employee class with bonus method
class Employee:
    company = "TechCorp"

    def __init__(self, name, monthly_salary, job_designation):
        self.name = name
        self.job_designation = job_designation
        self.__monthly_salary = monthly_salary  # Private: hides internal data
    
    def calculate_annual_salary(self):
        return self.__monthly_salary * 12
    
    def calculate_bonus(self, bonus_percentage=10):
        """
        Calculate bonus as a percentage of annual salary.
        Default is 10% if not specified.
        Why this approach? Bonus is calculated from annual salary,
        so it makes sense to put it in the Employee class methods.
        """
        return self.calculate_annual_salary() * (bonus_percentage / 100)
    
    def show_data(self):
        print(f"Company: {Employee.company}")
        print(f"Name: {self.name}")
        print(f"Designation: {self.job_designation}")
        print(f"Monthly Salary: ${self.__monthly_salary}")
        print(f"Annual Salary: ${self.calculate_annual_salary()}")
        print(f"Bonus (10%): ${self.calculate_bonus():.2f}")
        print("-" * 30)

# Test the enhanced Employee class
e1 = Employee("Alice Smith", 5000, "Software Engineer")
e2 = Employee("Bob Jones", 4620, "Data Scientist")
e3 = Employee("Charlie Brown", 4500, "QA Analyst")

e1.show_data()
e2.show_data()
e3.show_data()

# Demonstrate bonus with different percentages
print("Custom Bonus Calculation (15%):")
print(f"Alice's 15% bonus: ${e1.calculate_bonus(15):.2f}")
```

## Understanding Private Attributes & Adding a Bonus Method

**What is `__monthly_salary`?**
- The double underscore `__` prefix makes this attribute **private** (name mangled)
- Python converts `__monthly_salary` internally to `_Employee__monthly_salary` 
- This prevents accidental access from outside the class, protecting sensitive data
- You can't do `e1.__monthly_salary = 10000` directly — you'd get an AttributeError
- Why? **Encapsulation** — data integrity and controlled access through methods

**Why add a bonus method?**
- Keeps salary logic together in one place (Single Responsibility Principle)
- Bonus calculation depends on annual salary, so it belongs in the Employee class
- Makes the code reusable: any Employee object can calculate its own bonus

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        self.skills = []
    
    def show_info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
    
    def add_skills(self, skills):
        return self.skills.append(skills)
        
    
s1 = Student("Emily Davis", 20)
s2 = Student("Frank Miller", 22)

s1.skills('Java')
s2.skills('Python')
```
