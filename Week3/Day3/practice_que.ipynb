```python
my_list7_1= [1,2,3,4,5,4,3,2,1]
palindrome = True
for i in range(len(my_list7_1) // 2):
    if my_list7_1[i] != my_list7_1[-(i + 1)]:
        palindrome = False
        break
if palindrome == True:
    print("The my_list7_1 is a palindrome.")
else:
    print("The my_list7_1 is not a palindrome.")
```

```python
# Test Case 1 (Change this to test different lists)
my_list = ['r', 'a', 'd', 'a', 'r'] 

# Step 1: Initialize two pointers
left = 0
right = len(my_list) - 1

# Step 2: Create a flag variable to track if it's a palindrome
is_palindrome = True

# Step 3: Loop until the pointers meet in the middle
while left < right:
    if my_list[left] != my_list[right]:
        is_palindrome = False
        break  # Exit the loop early if any mismatch is found
        
    left += 1   # Move the left pointer forward
    right -= 1  # Move the right pointer backward

# Step 4: Output the final result
if is_palindrome:
    print("List is a Palindrome")
else:
    print("List is not a Palindrome")
```

```python
#Test with:
#● list1 = [1, 2, 3, 2, 1]
#● list2 = ['a', 'b', 'c', 'b', 'd']
#You are not allowed to use [::-1] for comparison.
def verify_palindrome(my_listx):
    palindrome = True
    for i in range(len(my_listx) // 2):
        if my_listx[i] !=  my_listx[-(i+1)]:
            palindrome = False
            break
        
    if palindrome == True:
        print(f"{my_listx} is a palindrome.")
    else:
        print(f"{my_listx} is not a palindrome.")
```

## Question 8

```python
#Given two lists:
#● list_a = [3, 1, 4]
#● list_b = [2, 5, 0]
my_list_a = [3, 1, 4]
my_list_b = [2, 5, 0]
my_list_c = []
my_list_c = my_list_a + my_list_b
print(f"Combined list: {my_list_c}")    
#bubble sort
for i in range(len(my_list_c)):
    for j in range(0, len(my_list_c) - i - 1):
        if my_list_c[j] > my_list_c[j + 1]:
            my_list_c[j], my_list_c[j + 1] = my_list_c[j + 1], my_list_c[j]

print(f"Sorted list: {my_list_c}")

    
```

## Question 9

```python
nested_list = [1, [2, 3], 4, [5, 6, 7], 8]
new_list = []
for item in nested_list:
    if type(item) == list:
        for sub_item in item:
            new_list.append(sub_item)
    else:
        new_list.append(item)

print(f"Flattened list: {new_list}")
```
