```python
# opening a file using context manager
with open("sample.txt") as f:
    data = f.read()
    print(data)
```

```python
# simple file copy operation
s = open("sample.txt", "r")
f = open("sm3.txt", "w")
```

```python
content = s.read()
f.write(content)
f.close()
s.close()
```
