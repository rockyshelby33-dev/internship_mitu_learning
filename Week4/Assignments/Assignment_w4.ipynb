## Question 1


```python
import numpy as np
import pandas as pd
###############################
def safe_divide(a, b):
    try:
        result = a / b
        print("Result:", result)

    except ZeroDivisionError:
        print("Error: Cannot divide by zero.")

    except TypeError:
        print("Error: Invalid data type.")

    finally:
        print("Operation attempted.")


# Testing
safe_divide(10, 2)
safe_divide(10, 0)
safe_divide(10, "abc")
```

## Question2


```python
class InvalidAgeError(Exception):
    pass


def validate_age(age):
    if age < 0 or age > 120:
        raise InvalidAgeError("Age must be between 0 and 120")
    return f"Valid age: {age}"


test_ages = [25, -5, 130]

for age in test_ages:
    try:
        print(validate_age(age))

    except InvalidAgeError as e:
        print("Invalid Age:", e)
```

## Question3

```python
user_inputs = ['10', 'abc', '0', '5']

for value in user_inputs:
    try:
        num = int(value)
        result = 100 / num
        print("Result:", result)

    except ValueError:
        print("ValueError: Input is not a number")

    except ZeroDivisionError:
        print("ZeroDivisionError: Cannot divide by zero")

    except Exception as e:
        print("Unexpected Error:", e)
```

## Question 6

```python
import numpy as np
import pandas as pd
np.random.seed(42)

arr = np.random.randint(1, 101, 15)

print("Array:", arr)

print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))

print("Descending Order:")
print(np.sort(arr)[::-1])

indices = np.argsort(arr)[-3:]
print("Indices of Top 3 Values:", indices)

normalized = (arr - arr.min()) / (arr.max() - arr.min())
print("Normalized Array:")
print(normalized)
```

## Question 7

```python
import pandas as pd

data = {
    "Name":["Aarav","Priya","Karan","Meera","Sahil"],
    "Age":[23,21,25,22,24],
    "Department":["AI","Data Science","AI","Web Dev","Data Science"],
    "Score":[88,92,79,95,67]
}

df = pd.DataFrame(data)

print(df.head(3))

print(df.dtypes)

def grade(score):
    if score >= 90:
        return "A"
    elif score >= 75:
        return "B"
    else:
        return "C"

df["Grade"] = df["Score"].apply(grade)

print(df.sort_values("Score", ascending=False))

print(df[df["Department"] == "AI"])
```

## Question 9

```python
import numpy as np
import pandas as pd

data = {
    "Product":["Pen","Notebook","Eraser","Sharpener","Ruler"],
    "Price":[10,np.nan,5,8,np.nan],
    "Stock":[100,50,np.nan,30,20]
}

products = pd.DataFrame(data)

print(products.isnull().sum())

mean_price = products["Price"].mean()

products["Price"].fillna(mean_price, inplace=True)

products = products.dropna(subset=["Stock"])

print(products)

products.to_csv("cleaned_products.csv", index=False)
```
