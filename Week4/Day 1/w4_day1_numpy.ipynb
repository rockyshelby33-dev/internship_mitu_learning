```python
# create two numpy arrays and perform element vise addition, subtraction, multiplication , division and modulus 
```

```python
array1 = np.array([10, 20, 30, 40, 50])
array2 = np.array([3,  4,  5,  6,  7])
addition = np.add(array1, array2)
print("np.add():       ", addition)
subtraction = np.subtract(array1, array2)
print("np.subtract():  ", subtraction)
multiplication = np.multiply(array1, array2)
print("np.multiply():  ", multiplication)
division = np.divide(array1, array2)
print("np.divide():    ", division)
#
modulus = np.mod(array1, array2)
print("np.mod():       ", modulus)
```
