```python
###################################
import numpy as np
import time

# Defining the exact 5 elements
l = [1, 2, 3, 4, 5]
ar = np.array([1, 2, 3, 4, 5]) # Corrected NumPy array syntax

# --- 1. Math Calculation on List ---
# Lists can't do math directly, so we must use a loop or list comprehension
start_list = time.perf_counter()
list_result = [x * 2 for x in l]
end_list = time.perf_counter()
list_time = end_list - start_list

# --- 2. Math Calculation on Array ---
# Arrays perform math instantly using vectorization
start_array = time.perf_counter()
array_result = ar * 2
end_array = time.perf_counter()
array_time = end_array - start_array

# --- 3. Results ---
print("List Result: ", list_result)
print("Array Result:", array_result)
print("-" * 40)
print(f"List Time:  {list_time:.8f} seconds")
print(f"Array Time: {array_time:.8f} seconds")
print(f"NumPy is {list_time / array_time:.2f} times faster at processing 5 elements.")
```
