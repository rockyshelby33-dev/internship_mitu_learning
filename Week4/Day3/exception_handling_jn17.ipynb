```python
try:
    f = open('a1.txt')
except FileNotFoundError:
    print("File does not exist")
except Exception as e:
    print(e)
else:
    print(f.read())
    f.close()
```

```python
#finally 
```

```python
try:
    f = open('a1.txt')
    print(var1)
except FileNotFoundError:
    print("File does not exist")
except Exception as e:
    print(e)
else:
    print(f.read())
    f.close()
finally:
    print("Run Complete")
```

```python
try:
    f = open('a1.txt')
    if f.name == 'a1.txt':
        raise Exception
except FileNotFoundError:
    print("File does not exist")
except Exception as e:
    print(e)
else:
    print(f.read())
    f.close()
finally:
    print("Run Complete")
```

```python
# take 2 numbers as user input 
n1 = int(input("Enter number1 "))
n2 = int(input("Enter number2 "))
#write a program for division of these 2 numbers
# handle ZeroDivisionError and ValueError
try:
    res_div = n1 / n2
    if res_div >= 0:
        print(res_div)
except ZeroDivisionError:
    print("cant divide by zero")
except ValueError:
    print("entered the wrong value")

finally:
    print("Division operation done")
```
