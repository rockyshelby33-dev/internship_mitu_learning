```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([10,23,29,39,56,63,72])
y = np.array([12,23,34,45,67,78,89])
z = np.array([89,79,68,67,56,45,34])
```

# Line Chart

```python
plt.plot(x, y)
```

```python
plt.xlabel("x Values")
plt.ylabel("Y Label")
plt.title("Line Chart: X vs Y")
plt.plot(x, y)
```

```python
plt.xlabel("x Values")
plt.ylabel("Y Label")
plt.title("Line Chart: X vs Y")
plt.plot(x, y, marker= 'o' ,mfc = 'red', ls = '--')
plt.grid()
```

```python
plt.xlabel("x Values")
plt.ylabel("Y Label")
plt.title("Line Chart: X vs Y")
plt.plot(x, y, marker= '*' ,mfc = 'red', ls = '--')
plt.grid()
```

```python
plt.xlabel("x Values")
plt.ylabel("Z Label")
plt.title("Line Chart: X vs Z")
plt.plot(x, z, marker= '*' ,mfc = 'green', ls = '--')
plt.grid()
```

```python
plt.xlabel("x Values")
plt.ylabel("Z Label")
plt.title("Line Chart: X vs Z")
plt.plot(x, z, marker= 'o' ,mfc = 'green', ls = '--')
plt.plot(x, y, marker= '*' ,mfc = 'red', ls = '--')
plt.grid()
```
