```python
# Visualize outliers using boxplot
plt.figure(figsize=(10, 5))
plt.boxplot(ads_df['Age'])
plt.title('Box Plot - Shows IQR and Outliers')
plt.ylabel('Age')
plt.grid()
plt.show()
```

```python
#  STATISTICAL ANALYSIS - IQR and Outliers
# ==========================================

# Step 1: Calculate Quartiles (Q1, Q3) and IQR
Q1 = ads_df['Age'].quantile(0.25)  # 25th percentile (first quartile)
Q3 = ads_df['Age'].quantile(0.75)  # 75th percentile (third quartile)
IQR = Q3 - Q1  # Interquartile Range

print("=== QUARTILE & IQR ANALYSIS ===")
print(f"Q1 (25th percentile): {Q1}")
print(f"Q3 (75th percentile): {Q3}")
print(f"IQR (Q3 - Q1): {IQR}")

# Step 2: Calculate Outlier Boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"\n=== OUTLIER BOUNDARIES ===")
print(f"Lower Bound: {lower_bound}")
print(f"Upper Bound: {upper_bound}")

# Step 3: Find Outliers
outliers = ads_df[(ads_df['Age'] < lower_bound) | (ads_df['Age'] > upper_bound)]
print(f"\n=== OUTLIERS FOUND ===")
print(f"Number of Outliers: {len(outliers)}")
print(f"Outlier Values: {outliers['Age'].values}")

# Step 4: Other Statistical Measures
print(f"\n=== OTHER STATISTICAL MEASURES ===")
print(f"Mean (Average): {ads_df['Age'].mean():.2f}")
print(f"Median (Middle Value): {ads_df['Age'].median():.2f}")
print(f"Mode (Most Frequent): {ads_df['Age'].mode().values}")
print(f"Std Dev (Spread): {ads_df['Age'].std():.2f}")
print(f"Min Value: {ads_df['Age'].min()}")
print(f"Max Value: {ads_df['Age'].max()}")
print(f"Variance: {ads_df['Age'].var():.2f}")
```

```python
# Step 1: Load the Tips Dataset
tips = sns.load_dataset('tips')

# Display first few rows to see the data
print("First 5 rows of Tips Dataset:")
tips.head()
```

```python
# Step 2: Basic Dataset Information
print("\n" + "="*50)
print("DATASET INFORMATION")
print("="*50)

print(f"Dataset Shape (Rows, Columns): {tips.shape}")
print(f"\nColumn Names and Types:")
print(tips.dtypes)
print(f"\nBasic Stats:")
tips.describe()
```

```python
# Chart 1: Distribution of Tips (Histogram)
# WHAT WE LEARN: How are tips distributed? Are most tips low or high?

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(tips['tip'], bins=15, color='skyblue', edgecolor='black')
plt.xlabel('Tip Amount ($)')
plt.ylabel('Frequency (Count)')
plt.title('Distribution of Tips\n(How many times each tip amount appears)')
plt.grid(alpha=0.3)

# Chart 2: Tip Percentage Distribution
tips['tip_percentage'] = (tips['tip'] / tips['total_bill']) * 100

plt.subplot(1, 2, 2)
plt.hist(tips['tip_percentage'], bins=15, color='lightcoral', edgecolor='black')
plt.xlabel('Tip Percentage (%)')
plt.ylabel('Frequency (Count)')
plt.title('Distribution of Tip Percentage\n(What % of bill is tipped?)')
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("\n📊 INSIGHT: Most tips are between $2-$5. Average tip percentage is around 15-20%")
```

```python
# Chart 2: Total Bill vs Tips (Scatter Plot)
# WHAT WE LEARN: Is there a relationship? Does higher bill = higher tip?

plt.figure(figsize=(10, 6))
plt.scatter(tips['total_bill'], tips['tip'], alpha=0.6, s=100, color='green', edgecolor='black')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip Amount ($)')
plt.title('Relationship: Total Bill vs Tips\n(Does higher bill lead to higher tips?)')
plt.grid(alpha=0.3)
plt.show()

print("\n📊 INSIGHT: YES! There's a positive relationship - higher bills usually get higher tips")
```

```python
# Chart 3: Tips by Day of Week (Bar Chart)
# WHAT WE LEARN: Which day gets the most tips? Are weekends different from weekdays?

plt.figure(figsize=(12, 5))

# Subplot 1: Average tip by day
plt.subplot(1, 2, 1)
day_tips = tips.groupby('day')['tip'].mean()
day_order = ['Thursday', 'Friday', 'Saturday', 'Sunday']
day_tips_sorted = day_tips.reindex(day_order)

plt.bar(day_tips_sorted.index, day_tips_sorted.values, color=['orange', 'red', 'purple', 'blue'], edgecolor='black')
plt.xlabel('Day of Week')
plt.ylabel('Average Tip ($)')
plt.title('Average Tips by Day\n(Which day has highest tips?)')
plt.xticks(rotation=45)
plt.grid(alpha=0.3, axis='y')

# Subplot 2: Total bill by day
plt.subplot(1, 2, 2)
day_bill = tips.groupby('day')['total_bill'].sum()
day_bill_sorted = day_bill.reindex(day_order)

plt.bar(day_bill_sorted.index, day_bill_sorted.values, color=['orange', 'red', 'purple', 'blue'], edgecolor='black')
plt.xlabel('Day of Week')
plt.ylabel('Total Bill Amount ($)')
plt.title('Total Bills by Day\n(Which day has most sales?)')
plt.xticks(rotation=45)
plt.grid(alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

print("\n📊 INSIGHT: Saturday has highest tips! Weekend is more profitable than weekdays")
```

```python
# Chart 4: Tips by Gender (Box Plot)
# WHAT WE LEARN: Do males and females tip differently?

plt.figure(figsize=(12, 5))

# Subplot 1: Box plot
plt.subplot(1, 2, 1)
tips.boxplot(column='tip', by='sex')
plt.xlabel('Gender')
plt.ylabel('Tip Amount ($)')
plt.title('Tip Distribution by Gender\n(Who tips more: Males or Females?)')
plt.suptitle('')  # Remove the automatic title

# Subplot 2: Bar chart for average tips
plt.subplot(1, 2, 2)
gender_tips = tips.groupby('sex')['tip'].agg(['mean', 'count'])
plt.bar(gender_tips.index, gender_tips['mean'], color=['skyblue', 'lightpink'], edgecolor='black')
plt.xlabel('Gender')
plt.ylabel('Average Tip ($)')
plt.title('Average Tips by Gender\n(Count shown on bars)')

for i, (idx, row) in enumerate(gender_tips.iterrows()):
    plt.text(i, row['mean'] + 0.1, f"${row['mean']:.2f}\nn={int(row['count'])}", ha='center')

plt.grid(alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

print("\n📊 INSIGHT: Males and females tip similarly on average")
print(f"Males average tip: ${tips[tips['sex']=='Male']['tip'].mean():.2f}")
print(f"Females average tip: ${tips[tips['sex']=='Female']['tip'].mean():.2f}")
```

```python
# Chart 5: Lunch vs Dinner (Time Analysis)
# WHAT WE LEARN: Is lunch or dinner more profitable? Who tips more?

plt.figure(figsize=(14, 5))

# Subplot 1: Count of transactions
plt.subplot(1, 3, 1)
time_count = tips['time'].value_counts()
colors = ['#FF6B6B', '#4ECDC4']
plt.bar(time_count.index, time_count.values, color=colors, edgecolor='black')
plt.xlabel('Time of Day')
plt.ylabel('Number of Transactions')
plt.title('Lunch vs Dinner\n(Which is busier?)')

for i, (idx, val) in enumerate(time_count.items()):
    plt.text(i, val + 1, str(val), ha='center')

# Subplot 2: Average tips
plt.subplot(1, 3, 2)
time_tips = tips.groupby('time')['tip'].mean()
plt.bar(time_tips.index, time_tips.values, color=colors, edgecolor='black')
plt.xlabel('Time of Day')
plt.ylabel('Average Tip ($)')
plt.title('Average Tips\n(Who tips more?)')

for i, (idx, val) in enumerate(time_tips.items()):
    plt.text(i, val + 0.05, f"${val:.2f}", ha='center')

# Subplot 3: Total sales
plt.subplot(1, 3, 3)
time_bills = tips.groupby('time')['total_bill'].sum()
plt.bar(time_bills.index, time_bills.values, color=colors, edgecolor='black')
plt.xlabel('Time of Day')
plt.ylabel('Total Sales ($)')
plt.title('Total Revenue\n(Which generates more money?)')

for i, (idx, val) in enumerate(time_bills.items()):
    plt.text(i, val + 20, f"${val:.0f}", ha='center')

plt.tight_layout()
plt.show()

print("\n📊 INSIGHT: Dinner is much busier than lunch and generates more revenue!")
```

```python
# Chart 6: Smoker vs Non-Smoker (Category Analysis)
# WHAT WE LEARN: Does smoking status affect tipping behavior?

plt.figure(figsize=(12, 5))

# Subplot 1: Average tips
plt.subplot(1, 2, 1)
smoker_tips = tips.groupby('smoker')['tip'].mean()
plt.bar(smoker_tips.index, smoker_tips.values, color=['green', 'red'], edgecolor='black')
plt.xlabel('Smoking Status')
plt.ylabel('Average Tip ($)')
plt.title('Average Tips by Smoking Status\n(Do smokers tip more?)')
plt.grid(alpha=0.3, axis='y')

for i, (idx, val) in enumerate(smoker_tips.items()):
    plt.text(i, val + 0.05, f"${val:.2f}", ha='center')

# Subplot 2: Scatter - Bill vs Tip colored by smoker
plt.subplot(1, 2, 2)
for smoker_status in ['Yes', 'No']:
    mask = tips['smoker'] == smoker_status
    color = 'red' if smoker_status == 'Yes' else 'green'
    label = 'Smokers' if smoker_status == 'Yes' else 'Non-Smokers'
    plt.scatter(tips[mask]['total_bill'], tips[mask]['tip'], 
               alpha=0.6, s=80, label=label, color=color, edgecolor='black')

plt.xlabel('Total Bill ($)')
plt.ylabel('Tip Amount ($)')
plt.title('Bill vs Tips by Smoking Status')
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("\n📊 INSIGHT: Smokers and non-smokers have similar tipping patterns")
print(f"Smokers average: ${tips[tips['smoker']=='Yes']['tip'].mean():.2f}")
print(f"Non-Smokers average: ${tips[tips['smoker']=='No']['tip'].mean():.2f}")
```

```python
# Chart 7: Party Size vs Tips
# WHAT WE LEARN: Do larger groups tip more or less?

plt.figure(figsize=(12, 5))

# Subplot 1: Box plot
plt.subplot(1, 2, 1)
tips.boxplot(column='tip', by='size')
plt.xlabel('Party Size (Number of People)')
plt.ylabel('Tip Amount ($)')
plt.title('Tips by Party Size\n(Do bigger groups tip more?)')
plt.suptitle('')  # Remove automatic title

# Subplot 2: Average tips and bill
plt.subplot(1, 2, 2)
size_analysis = tips.groupby('size').agg({'tip': 'mean', 'total_bill': 'mean'})
plt.bar(size_analysis.index, size_analysis['tip'], color='skyblue', edgecolor='black', label='Avg Tip')
plt.xlabel('Party Size')
plt.ylabel('Average Amount ($)')
plt.title('Average Tips by Party Size')
plt.legend()
plt.grid(alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

print("\n📊 INSIGHT: Larger parties tip more in absolute dollars but similar percentage")
```
