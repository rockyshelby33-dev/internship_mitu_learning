```python
from sklearn.preprocessing import LabelEncoder

# define a reusable function for label encoding a DataFrame
# 1) copy the DataFrame so we do not change the original data in place
# 2) drop the pickup and dropoff columns as requested
# 3) find all categorical columns and convert them to numeric labels

def label_encode_df(df):
    df = df.copy()
    df = df.drop(columns=['pickup', 'dropoff'], errors='ignore')

    cat_columns = df.select_dtypes(include=['object', 'category']).columns.tolist()
    le = LabelEncoder()

    for col in cat_columns:
        df[col] = le.fit_transform(df[col].astype(str))

    return df

# apply the function to the taxi dataset
taxi_encoded = label_encode_df(df)
print(taxi_encoded.head())
print('\nEncoded columns:', taxi_encoded.select_dtypes(include=['int64', 'float64']).columns.tolist())
```
