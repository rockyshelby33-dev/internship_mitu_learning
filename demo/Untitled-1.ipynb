```python
def approve_loan(income, credit_score, existing_loan):

    if income >= 5000 and credit_score >= 700 and not existing_loan:
        return "Loan Approved"
    elif income >= 3000 and credit_score >= 650 :
        return "Further Review Required"
    else:
        return "Loan Rejected"
```

```python
income = 700000
credit_score = 750
existing_loan = False
```

```python
# Interfereance Engine
```

```python
decision = approve_loan(
    income,
    credit_score,
    existing_loan
)
```
